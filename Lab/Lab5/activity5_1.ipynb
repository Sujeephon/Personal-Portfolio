1.คุณจะนำ Dictionary ไปใช้ตรงไหนในเกมได้บ้าง? (เช่น ช่องเก็บของของผู้เล่น (Inventory), ค่าพลังของตัวละคร)
    ใช้สำหรับเก็บข้อมูลผู้เล่น เช่น ข้อมูลตัวละคร ได้แก่ ชื่อ, เลเวล, HP, MP และคะแนน
2.Dictionary จะช่วยจัดเก็บข้อมูลสินค้าในร้านค้าออนไลน์ได้อย่างไร? (เช่นข้อมูลรายละเอียดสินค้าโดยใช้อ้างอิงจากรหัสสินค้า/ID)
    ใช้เพื่อเก็บรายละเอียดของสินค้า โดยใช้รหัสสินค้าเป็น Key และเก็บข้อมูลสินค้า เช่น ชื่อ ราคา จำนวนสินค้า และหมวดหมู่เป็น Value
3.Set จะมีประโยชน์อย่างไรในการจัดการสิทธิ์ของผู้ใช้ หรือการจัดการแท็ก? (เช่น บทบาทสิทธิ์ใช้งานที่ไม่ซ้ำกัน, แฮชแท็กที่ไม่ซ้ำกัน)
    ใช้สำหรับเก็บสิทธิ์ของผู้ใช้ เช่น admin, editor, viewer หรือใช้เก็บ Hashtag ของโพสต์
4.นึกออกไหมว่ามีสถานการณ์ไหนบ้างที่เราต้องเช็กอย่างรวดเร็วว่า "มีข้อมูลนี้อยู่ในกลุ่มข้อมูลขนาดใหญ่หรือไม่" โดยไม่สนใจลำดับก่อนหลัง? (เช่น รายชื่อ IP Address ที่ถูกบล็อก, การนับจำนวนผู้เข้าชมเว็บที่ไม่นับคนซ้ำ)
    ใช้สำหรับตรวจสอบว่าข้อมูลมีอยู่ในระบบหรือไม่ เช่น ตรวจสอบว่า IP Address อยู่ในรายการที่ถูกบล็อกหรือไม่
